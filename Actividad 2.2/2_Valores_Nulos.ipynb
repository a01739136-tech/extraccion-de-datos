```python
#Carga desde un archivo .xlsx sin indice
df= pd.read_excel('Ventas_por_Asesor.xlsx')
df.head(5)
```

```python
df.columns
```

```python
#Se revisó la base en excel, en excel se dividió la base en dos partes. De manera mensual y de manera anual. Se limpiarán ambos.
```

```python

# Leer únicamente la pestaña de ventas_por_asesor_anual
df_anual = pd.read_excel("Ventas_por_Asesor.xlsx", sheet_name="ventas_por_asesor_anual")
```

```python
df_anual.head(5)
```

```python
#Identificar valores nulos por columna
valores_nulos_anual=df_anual.isnull().sum()
valores_nulos_anual
```

```python
#Se cambian los nulos por 0, ya que el 0 no modifica el resultado de la suma de ventas, y se puede considerar que no hubo ventas en ese año.
df_anual.fillna(0, inplace=True).head(5)
```

```python
#Corroborar que no hay valores nulos
valores_nulos_anual=df_anual.isnull().sum()
valores_nulos_anual
```

```python
# Guardarla como un csv independiente
df_anual.to_csv("Ventas_A_Anual_Limpia.csv", index=False)
```

```python
# Leer únicamente la pestaña de ventas_por_asesor_mensual
df_mensual = pd.read_excel("Ventas_por_Asesor.xlsx", sheet_name="ventas_por_asesor_mensual")
```
