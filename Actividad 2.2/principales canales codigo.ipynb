```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .xlsx sin indice
df= pd.read_excel('Principales_Canales.xlsx')
df.head(5)
```

```python
#Se observa que en el excel hay distintas tablas, por lo que se separarán en distintos dataframes, para poder trabajar con ellos de manera independiente. Se separan las tablas de Piso, Plaza y Digital. No se tomará la última tabla de digital debido a que no tiene información relevante para el análisis.
```

```python
#Piso
#Data frame de la tabla de piso, se toma desde la fila 5 hasta la 14 y de la columna 0 a la 31, ya que es donde se encuentra la información relevante.
df_piso = df.iloc[4:14, 0:31].copy()
df_piso
```

```python
#Se pone como encabezado la primera fila del dataframe para un mejor análisis.
df_piso.columns = df_piso.iloc[0]
df_piso = df_piso.iloc[1:].reset_index(drop=True)

df_piso
```

```python
#Limpieza de datos, los valores nulos se reemplazan por NaN, ya que no se puede considerar que el valor fue 0 por que se interpretaría como que ocurrió y el resultado fue 0, en vez de que no hay registro de ello.
df_piso.fillna(np.nan, inplace=True)
```

```python
#Identificar valores nulos por columna
valores_nulos=df_piso.isnull().sum()
valores_nulos
```

```python
#Plaza
#Data frame de la tabla de plaza, se toma desde la fila 18 hasta la 29 y de la columna 0 a la 31, ya que es donde se encuentra la información relevante. Y se hace una copia del mismo para no modificar el original.
df_plaza = df.iloc[17:29, 0:31].copy()
df_plaza

df_plaza.columns = df_plaza.iloc[0]
df_plaza = df_plaza.iloc[1:].reset_index(drop=True)

df_plaza
```

```python
#Identificar valores nulos por columna
valores_nulos=df_plaza.isnull().sum()
valores_nulos
```

```python
#Limpieza de datos, los valores nulos se reemplazan por NaN, ya que no se puede considerar que el valor fue 0 por que se interpretaría como que ocurrió y el resultado fue 0, en vez de que no hay registro de ello.
df_plaza.fillna(np.nan, inplace=True)
```

```python
#Corroborar que los valores nulos fueron reemplazados por NaN
valores_nulos=df_plaza.isnull().sum()
valores_nulos
```

```python
#Digital
df_digital = df.iloc[32:42, 0:31].copy()
df_digital

df_digital.columns = df_digital.iloc[0]
df_digital = df_digital.iloc[1:].reset_index(drop=True)

df_digital
```

```python
#Identificar valores nulos por columna
valores_nulos=df_digital.isnull().sum()
valores_nulos
```

```python
#Limpieza de datos, los valores nulos se reemplazan por NaN, ya que no se puede considerar que el valor fue 0 por que se interpretaría como que ocurrió y el resultado fue 0, en vez de que no hay registro de ello.
df_digital.fillna('NaN', inplace=True)
```

```python
#Corroborar que los valores nulos fueron reemplazados por NaN
valores_nulos=df_digital.isnull().sum()
valores_nulos
```

```python
import numpy as np

for dataframe in [df_piso, df_plaza, df_digital]:
    dataframe.replace("-", np.nan, inplace=True)
```

```python
#Exportar los dataframes en un documento limpios a archivos .csv para su posterior análisis
df_piso.to_csv("Ventas_Piso_Mensual_Limpia.csv", index=False)
df_plaza.to_csv("Ventas_Plaza_Mensual_Limpia.csv", index=False)
df_digital.to_csv("Ventas_Digital_Mensual_Limpia.csv", index=False)
```
