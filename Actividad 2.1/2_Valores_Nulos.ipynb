```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('Citas_Digital_Filtrado (1).csv')
data.head(5)
```

```python
#Eliminar columnas innecesarias
#axis = 1 para columnas
#axis = 0 para filas
data1= data.drop(['Unnamed: 15'], axis=1) #el axis = 0 se refiere a las filas y con 1 es columnas
data1
```

```python
#Identificar valores nulos por columna
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

**Métodos de Sustitución de Valores Nulos**

```python
data2=data1.copy()
```

```python
#BDC
#Sustituir valores nulos por valores no nulos hacia adelante  "forward fill" ("ffill"), parte desde el valor cero, el primero, hasta abajo. Y cuando hayan valores nulos lo reemplaza con el que estaba arriba
#Filtro por columnas
data2["BDC "] =data2["BDC "].ffill()
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#FECHA
#Sustituir valores nulos por un string en  concreto, vamos a sustituir los valores nulos de la columna "Fecha Venta" por un string en concreto, en este caso "2030-06-19 00:00:00"
data2["Fecha"] =data2["Fecha"].fillna("2030-06-19 00:00:00") 
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#NOMBRE CLIENTE
#Explicación: vamos a sustituir los valores nulos de la columna "Nombre Cliente" por un string en concreto, en este caso "Sin Nombre"
data2["Nombre Cliente "] =data2["Nombre Cliente "].fillna("Sin Nombre")
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Teléfono
#Explicación: vamos a sustituir los valores nulos de la columna "Teléfono" por un string en concreto, en este caso "Sin Teléfono"
data2["Telefono "] =data2["Telefono "].fillna("Sin Telefono")
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#ESTATUS LEAD
#Explicación: vamos a sustituir los valores nulos de la columna "Estatus Lead" por un string en concreto, en este caso "Sin Estatus"
data2["Estatus de Lead"] =data2["Estatus de Lead"].fillna("Sin Estatus")
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#SE revisan todos los tipos de datos que hay en esta columa
data2['Potencial de compra'].unique()
```

```python
#Se reemplazan los valores de la columna "Potencial de compra" para poder convertirlos a tipo float
data2['Potencial de compra'].unique

data2['Potencial de compra']=data2['Potencial de compra'].astype("string")
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("Ventas ","0", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("%","", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("'","", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("&","", regex=False)

data2['Potencial de compra']=data2['Potencial de compra'].astype(float)
```

```python
#Potencial de compra
#Explicación vamos a sustituir los valores nulos de la columnas pro la mediana de la columna, en este caso "Potencial de Compra"
data2["Potencial de compra"] = data2["Potencial de compra"].fillna(round(data2["Potencial de compra"].median(),1))
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Venta
#Los valores nulos se sustiyen con el número 2, que tendra en significado de "No se tiene registro de venta"
data2["Venta"] =data2["Venta"].fillna(2)
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Asesor asigando
#Explicación: vamos a sustituir los valores nulos de la columna "Asesor asignado" por un string en concreto, en este caso "Sin Asesor"
data2["Asesor Asignado"] = data2["Asesor Asignado"].fillna("Sin Asesor")
```

```python
#Corroborar
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```
